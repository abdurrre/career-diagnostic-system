```python
def assign_job_category(title):
    """
    Memetakan job_title ke 7 kategori profesi utama.
    Profesi di luar 7 target utama akan dipetakan ke 'Other IT'.
    """
    if pd.isna(title):
        return 'Other IT'
    t = title.lower()

    # 1. Antisipasi False Positives (Cek ini di awal)
    if 'data loss prevention' in t or 'dlp' in t:
        return 'Other IT' # Atau 'Security' jika ingin tetap dianalisis

    # 2. AI / Machine Learning Engineer
    if any(k in t for k in [
        'machine learning', 'ml engineer', 'ai engineer', 'artificial intelligence',
        'deep learning', 'nlp', 'computer vision', 'mlops'
    ]):
        return 'AI / Machine Learning Engineer'

    # 3. Data Scientist
    if any(k in t for k in ['data scientist', 'data science', 'research scientist']):
        return 'Data Scientist'

    # 4. Data Engineer
    if any(k in t for k in [
        'data engineer', 'data architect', 'database administrator', 'dba',
        'analytics engineer', 'big data'
    ]):
        return 'Data Engineer'

    # 5. Data Analyst
    if any(k in t for k in [
        'data analyst', 'bi developer', 'business intelligence', 'data visualization'
    ]):
        return 'Data Analyst'

    # 6. Fullstack Developer
    if any(k in t for k in ['full stack', 'fullstack', 'full-stack']):
        return 'Fullstack Developer'

    # 7. Frontend Developer
    if any(k in t for k in [
        'frontend', 'front-end', 'front end', 'ui developer', 'react developer',
        'vue developer', 'angular developer'
    ]):
        return 'Frontend Developer'

    # 8. Backend Developer
    if any(k in t for k in [
        'backend', 'back-end', 'back end', 'software engineer', 'software developer',
        'python developer', 'java developer', 'php developer', 'node developer',
        'golang', 'api developer'
    ]):
        return 'Backend Developer'

    # Jika tidak masuk 7 role di atas, buang ke Other IT
    return 'Other IT'


df_final['job_category'] = df_final['job_title'].apply(assign_job_category)

print('Distribusi job_category:')
print(df_final['job_category'].value_counts())
print(f'\nTotal baris: {len(df_final)}')
```

```python
# Memfilter dataframe untuk membuang 'Other IT'
df_final_filtered = df_final[df_final['job_category'] != 'Other IT'].copy()

# Mengecek distribusi data setelah difilter
print("=== Distribusi 7 Role Capstone ===")
print(df_final_filtered['job_category'].value_counts())

# Mengecek total baris yang tersisa
print(f'\nTotal baris awal: {len(df_final)}')
print(f'Total baris sekarang (setelah Other IT di-drop): {len(df_final_filtered)}')
```
